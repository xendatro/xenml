```python
import numpy as np
import sys
import matplotlib.pyplot as plt
import sklearn

sys.path.insert(0, "..")
from regressors.LinearRegression import LinearRegression
```

```python
diabetes = sklearn.datasets.load_diabetes()
data = diabetes['data']
target = diabetes['target']

model = LinearRegression()

model.fit(data, target)
predictions = model.predict(data)

plt.scatter(target, predictions, alpha=0.6, edgecolors="k", linewidths=0.3)
plt.plot([target.min(), target.max()], [target.min(), target.max()], "r--", label="perfect prediction")
plt.xlabel("Actual progression")
plt.ylabel("Predicted progression")
plt.title("Obesity progression predicted via lin. regression vs. actual")
plt.legend()
plt.gca().set_aspect("equal")
plt.show()
```
